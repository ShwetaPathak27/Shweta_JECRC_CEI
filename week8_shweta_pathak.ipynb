```python
# TOOL 1: Calculator
def calculator(expression: str) -> str:
    """Evaluate a mathematical expression."""
    try:
        return str(eval(expression))
    except Exception:
        return "Error in calculation"

#TOOL 2: Keyword Extractor
def extract_keywords(text: str) -> list:
    """Extract keywords from text ."""
    try:
        words = text.split()
        # Filter words longer than 4 characters and lowers them
        keywords = list(set([w.lower() for w in words if len(w) > 4]))
        return keywords[:5]
    except Exception:
        return []

```

```python
# Implement the Agentic Logic

import json
import re

def agent(query: str):
    query_lower = query.lower()

    #Task-routing logic based on keywords
    if "calculate" in query_lower or re.search(r'[\+\-\*/]', query_lower):
        # Extract the expression part (everything after calculate)
        expression = query_lower.replace("calculate", "").strip()
        result =  calculator(expression)

        # Check if calculator returned an error
        if result == "Error in calculation":
            return {"type": "error", "result": result}
        else:
            return {"type": "calculation", "result": result}

    elif "keywords" in query_lower:
        # Extract the text part (everything after extract)
        text_content = query_lower.split("keywords")[-1].strip()
        result = extract_keywords(text_content)
        return {"type": "keywords", "result": result}

    else:
        # Fallback for general conversation
        return {"type": "general",
                "result": "I am a smart assistant. I can help with calculations or keyword extraction."
        }

```

```python
# Test case
queries = [
    "Calculate 20 + 5",
    "Extract keywords from Artificial Intelligence is transforming industries",
    "What is the weather?"
]

for q in queries:
    print(f"Query: {q}")
    print(f"Response: {agent(q)}")
    print("-" * 50)
```

```python
# Interactive Mode
import json

#Interactive Mode
print("Agent is ready! (Type 'exit' to stop)")
while True:
    user_input = input("Enter query: ")
    if user_input.lower() == "exit":
        break

    # Get the responses and print it as clean json like structure
    response = agent(user_input)
    print(json.dumps(response, indent=2))
```
